```
from datetime import datetime

# 1. Configuração inicial das variáveis e base de respostas
historico = []
RESPOSTAS = {
    'nome': 'Meu nome é ByteBot, seu assistente digital!',
    'ajuda': 'Posso responder sobre: nome, turno, historico, hora.',
    'hora': lambda: f'Agora são {datetime.now().strftime("%H:%M:%S")}.'
}

# 2. Lógica de processamento (com a correção da ajuda inclusa)
def processar_mensagem(mensagem: str, turno: int) -> str:
    msg = mensagem.lower().strip()

    if 'nome' in msg:
        return RESPOSTAS['nome']
    elif 'hora' in msg:
        return RESPOSTAS['hora']()
    elif 'ajuda' in msg:
        return RESPOSTAS['ajuda']
    elif 'turno' in msg:
        return f"Já conversamos por {turno} turnos até agora."
    elif 'historico' in msg or 'histórico' in msg:
        # Pega as últimas 3 mensagens do usuário guardadas na lista global
        ultimas = [t['usuario'] for t in historico[-3:]]
        if ultimas:
            return f"Suas últimas mensagens foram: {', '.join(ultimas)}"
        else:
            return "Ainda não temos mensagens registradas no histórico."
    else:
        return 'Não entendi. Digite "ajuda" para ver o que posso fazer.'

# 3. Função para salvar o contexto da conversa
def registrar_turno(turno: int, usuario: str, bot: str):
    historico.append({
        'turno': turno,
        'usuario': usuario,
        'bot': bot,
        'timestamp': datetime.now()
    })

# 4. Interface principal do chatbot
def chatbot_contextual():
    print('=== ByteBot - Chatbot com Memória ===')
    print('Dica: Tente falar "nome", "hora", "ajuda" ou "histórico".')
    print('Digite "sair" para encerrar.\n')

    global historico
    historico = [] # Limpa o histórico ao iniciar uma nova sessão
    turno = 0

    while True:
        entrada = input('Você: ').strip()

        if not entrada:
            continue

        if entrada.lower() in ('sair', 'tchau', 'encerrar'):
            print(f'\nBot: Até mais! Foram {turno} turnos de conversa hoje.')
            break

        turno += 1

        # Gera a resposta baseada na lógica
        resposta = processar_mensagem(entrada, turno)

        # Salva no histórico para consulta futura (Memória de Contexto)
        registrar_turno(turno, entrada, resposta)

        print(f'Bot: {resposta}\n')

# 5. Execução
if __name__ == "__main__":
    chatbot_contextual()
```

```
import re

# Listas de referência (Entidades)
SABORES = ['calabresa', 'frango', 'queijo', 'portuguesa', 'vegetariana', 'pepperoni']
TAMANHOS = ['pequena', 'media', 'média', 'grande', 'gigante', 'familia']

def detectar_intencao(mensagem: str) -> str:
    msg = mensagem.lower()
    if any(p in msg for p in ['quero', 'pedir', 'pedido', 'comprar', 'queria']):
        return 'FAZER_PEDIDO'
    elif any(p in msg for p in ['cancelar', 'desistir', 'não quero']):
        return 'CANCELAR'
    elif any(p in msg for p in ['cardápio', 'cardapio', 'opções', 'opcoes', 'tem']):
        return 'VER_CARDAPIO'
    return 'DESCONHECIDO'

# 1. IMPLEMENTAÇÃO: extrair_entidades()
def extrair_entidades(mensagem: str) -> dict:
    msg = mensagem.lower()
    entidades = {'sabor': None, 'tamanho': None}

    # Busca por sabores na frase
    for sabor in SABORES:
        if sabor in msg:
            entidades['sabor'] = sabor

    # Busca por tamanhos na frase
    for tamanho in TAMANHOS:
        if tamanho in msg:
            entidades['tamanho'] = tamanho

    return entidades

# 2. IMPLEMENTAÇÃO: confirmar_pedido()
def confirmar_pedido(entidades: dict) -> str:
    sabor = entidades.get('sabor')
    tamanho = entidades.get('tamanho')

    if sabor and tamanho:
        # Corrige a exibição da palavra 'média' caso o usuário digite sem acento
        tamanho_fmt = "Média" if tamanho in ['media', 'média'] else tamanho.capitalize()
        return f"Confirmado! Uma pizza {tamanho_fmt} de {sabor.capitalize()} saindo para entrega!"
    elif sabor:
        return f"Pizza de {sabor.capitalize()}! Excelente escolha. Mas qual o tamanho? (Pequena, Média ou Grande?)"
    elif tamanho:
        return f"Uma pizza {tamanho.capitalize()}, entendi. Mas qual o sabor você deseja?"
    else:
        return "Ainda não consegui entender o sabor e o tamanho. Pode repetir seu pedido?"

def chatbot_pizza():
    print('=== PizzaBot - Sistema de Pedidos ===')
    print('Dica: "Quero uma pizza grande de calabresa" ou "Ver cardápio"')

    while True:
        entrada = input('Você: ').strip()
        if entrada.lower() in ('sair', 'tchau'):
            print('Bot: Até a próxima!')
            break

        intencao = detectar_intencao(entrada)
        print(f' [DEBUG] Intenção: {intencao}')

        if intencao == 'VER_CARDAPIO':
            print(f'Bot: Sabores: {", ".join(SABORES)}')
            print(f'Bot: Tamanhos: Pequena, Média, Grande, Gigante e Família.')
        elif intencao == 'FAZER_PEDIDO':
            entidades = extrair_entidades(entrada)
            print(f' [DEBUG] Entidades: {entidades}')
            resposta = confirmar_pedido(entidades)
            print(f'Bot: {resposta}')
        elif intencao == 'CANCELAR':
            print('Bot: Sem problemas, pedido cancelado.')
        else:
            print('Bot: Não entendi. Você pode pedir uma pizza ou ver o cardápio.')
        print()

# Executar
chatbot_pizza()
```

```
import re

# 1. IMPLEMENTAÇÃO: Classe Personalidade
class Personalidade:
    def __init__(self, nome, tom, prefixo, vocab_proibido):
        self.nome = nome
        self.tom = tom
        self.prefixo = prefixo
        self.vocab_proibido = vocab_proibido

# Definição das personalidades disponíveis
PERSONALIDADES = {
    'formal': Personalidade(
        nome='Assistente Formal',
        tom='cordial e profissional',
        prefixo='Prezado usuário, ',
        vocab_proibido=['cara', 'mano', 'oi', 'beleza', 'vlw']
    ),
    'casual': Personalidade(
        nome='ChatZinho',
        tom='descontraído e jovem',
        prefixo='Oi! ',
        vocab_proibido=['prezado', 'solicito', 'outrossim', 'conforme']
    ),
    'tecnico': Personalidade(
        nome='TechBot',
        tom='técnico e direto',
        prefixo='RESPOSTA: ',
        vocab_proibido=['acho', 'talvez', 'quem sabe']
    )
}

RESPOSTAS_GENERICAS = {
    'python': 'Python é uma linguagem de programação interpretada e de alto nível.',
    'chatbot': 'Chatbot é um sistema que simula conversas humanas de forma automatizada.',
    'ia': 'Inteligência Artificial é a capacidade de máquinas realizarem tarefas cognitivas.',
    'ajuda': 'Posso falar sobre: python, chatbot, ia. Troque personalidade com /modo.'
}

class ChatbotMultiPersonalidade:
    def __init__(self):
        # Inicia com a personalidade formal por padrão
        self.personalidade_ativa = PERSONALIDADES['formal']
        self.historico = []

    # 2. IMPLEMENTAÇÃO: trocar_personalidade()
    def trocar_personalidade(self, nome: str) -> str:
        if nome in PERSONALIDADES:
            self.personalidade_ativa = PERSONALIDADES[nome]
            return f"Personalidade alterada para: {self.personalidade_ativa.nome} (Tom: {self.personalidade_ativa.tom})"
        return "Personalidade não encontrada. Tente: formal, casual ou tecnico."

    # 3. IMPLEMENTAÇÃO: gerar_resposta()
    def gerar_resposta(self, mensagem: str) -> str:
        msg = mensagem.lower()

        # 1. Busca a resposta base
        resposta_base = "Não tenho informações específicas sobre isso. Tente perguntar sobre Python ou IA."
        for chave, texto in RESPOSTAS_GENERICAS.items():
            if chave in msg:
                resposta_base = texto
                break

        # 2. Aplica o prefixo da personalidade ativa
        resposta_final = self.personalidade_ativa.prefixo + resposta_base

        # 3. Filtra vocabulário proibido (substitui por ***)
        for termo in self.personalidade_ativa.vocab_proibido:
            # Usando re para garantir que substitua independente de maiúsculas/minúsculas
            padrao = re.compile(re.escape(termo), re.IGNORECASE)
            resposta_final = padrao.sub("***", resposta_final)

        return resposta_final

    def executar(self):
        print('=== Chatbot Multi-Personalidade ===')
        print('Comandos: /modo formal | /modo casual | /modo tecnico | sair')
        print(f'Personalidade ativa: {self.personalidade_ativa.nome}\n')

        while True:
            entrada = input('Você: ').strip()
            if not entrada: continue

            if entrada.lower() == 'sair':
                print('Bot: Encerrando. Até logo!')
                break

            elif entrada.lower().startswith('/modo'):
                try:
                    nome_modo = entrada.split(' ', 1)[1].lower()
                    print(f'Bot: {self.trocar_personalidade(nome_modo)}\n')
                except IndexError:
                    print('Bot: Por favor, especifique um modo (ex: /modo casual).\n')
            else:
                resposta = self.gerar_resposta(entrada)
                print(f'Bot: {resposta}\n')

# 4. TESTE: Executando o bot
if __name__ == "__main__":
    bot = ChatbotMultiPersonalidade()
    bot.executar()
```
